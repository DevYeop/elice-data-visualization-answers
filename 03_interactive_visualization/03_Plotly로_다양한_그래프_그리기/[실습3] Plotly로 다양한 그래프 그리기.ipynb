# [실습3] Plotly 로 다양한 그래프 그리기

---

이번 시간에는 Plotly에서 제공하는 다양한 유형의 그래프를 그려보도록 하겠습니다. 

이전 실습들과 동일하게 필요한 라이브러리들을 불러옵니다. 

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

이번 시간에 그려볼 그래프의 유형은 다음과 같습니다.

1) 산점도
2) 선 그래프
3) 막대 그래프
4) 파이 그래프
5) 박스 플롯
6) 히스토그램
7) 히트맵
8) 타임라인 / 간트 차트
9) 퍼널 차트
10) 지도

위 그래프들을 우리는 크게 2가지 방식 (Express, Graph Object) 으로 알아볼 예정입니다. 

## 1. 산점도

산점도 그래프는 두 변수 간의 관계를 시각화하는 데 사용되는 통계적 그래프로, 각각의 데이터 포인트는 그래프의 x축과 y축 상에서 위치합니다.

산점도 그래프를 통해 우리는 두 변수의 상관관계를 직관적으로 확인할 수 있습니다.

**Plotly Express 버전 코드**

```python
fig = px.scatter(x=[0, 1, 2, 3, 4], y=[0, 1, 4, 9, 16])

fig.show()
```

**Graph Object 버전 코드**

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=[0, 1, 2, 3, 4], y=[0, 1, 4, 9, 16], mode="markers"))

fig.show()
```

## 2. 선 그래프

선 그래프는 가로축과 세로축을 갖는 좌표 평면 상에 데이터를 선으로 표현한 그래프입니다. 

막대 그래프와 함께 복수의 데이터를 비교하는데 유용한 시각화 방법이며, 시간, 순서, 변화 등에 따른 데이터의 추이와 패턴을 시각적으로 파악하기에 유용합니다. 

**Plotly Express 버전 코드**

```python
df = px.data.gapminder().query("continent == 'Oceania'")
fig = px.line(df, x='year', y='lifeExp', color='country')

fig.show()
```

**Graph Object 버전 코드**

```python
df = px.data.gapminder().query("continent == 'Oceania'")
fig = go.Figure()

# 각 나라별로 데이터를 그룹화하여 라인 추가
for country in df['country'].unique():
    country_data = df[df['country'] == country]
    fig.add_trace(go.Scatter(x=country_data['year'], y=country_data['lifeExp'], mode='lines', name=country))
fig.show()
```

## 3. 막대 그래프

막대 그래프는 가로축과 세로축을 갖는 좌표 평면 상에 데이터를 막대로 표현한 그래프, 값의 크고 작음을 한눈에 파악하기 용이한 시각화 형태입니다. 

**Plotly Express 버전 코드**

```python
animals = ['giraffes', 'orangutans', 'monkeys']
counts = [20, 14, 23]
fig = px.bar(x=animals, y=counts)
fig.show()
```

**Graph Object 버전 코드**

```python
animals=['giraffes', 'orangutans', 'monkeys']
fig = go.Figure([go.Bar(x=animals, y=[20, 14, 23])])
fig.show()
```

## 4. 파이 그래프

파이 차트는 전체 데이터에서 그 범주에 해당하는 비율을 표현하는 그래프로, 원형으로 표현되며, 부채꼴의 중심각은 해당 범주의 상대적인 비율을 표현합니다.

**Plotly Express 버전 코드**

```python
labels = ['Oxygen', 'Hydrogen', 'Carbon_Dioxide', 'Nitrogen']
values = [4500, 2500, 1053, 500]
fig = px.pie(names=labels, values=values)
fig.show()
```

**Graph Object 버전 코드**

```python
labels = ['Oxygen','Hydrogen','Carbon_Dioxide','Nitrogen']
values = [4500, 2500, 1053, 500]
fig = go.Figure(data=[go.Pie(labels=labels, values=values)])
fig.show()
```

## 5. 박스 플롯

박스 플롯은 데이터의 분포를 보여주는 시각화 그래프로, 수치형 변수의 데이터를 요약하고 비교하는 데 사용합니다. 

단순하게 평균을 보면 알 수 없는 데이터의 중앙값과 이상치, 분포를 파악할 수 있습니다.

**Plotly Express 버전 코드**

```python
df = px.data.tips()
fig = px.box(df, y="total_bill")
fig.show()
```

**Graph Object 버전 코드**

```python
df = px.data.tips()
fig = go.Figure(data=[go.Box(y=df['total_bill'])])
fig.show()
```

## 6. 히스토그램

히스토그램은 특정 구간에 해당하는 빈도를 표현하는 그래프로, 이 구간에 속한 데이터는 몇 개가 있는지를 표현합니다. 

특정 값의 빈도 분포를 분석할 때 유용하게 사용할 수 있습니다.

**Plotly Express 버전 코드**

```python
df = px.data.tips()
fig = px.histogram(df, x="total_bill")
fig.show()
```

**Graph Object 버전 코드**

```python
df = px.data.tips()
fig = go.Figure(data=[go.Histogram(x=df['total_bill'])])
fig.show()
```

## 7. 히트맵

히트맵은 데이터 값의 상대적인 크기를 색상으로 표시하여 시각화하는 그래프로, 2차원 표 형태의 데이터를 사용하여 생성합니다.

값의 크기에 따라 다른 색상 맵을 사용하며, 값이 클수록 진한 색, 작을수록 연한 색으로 표현되는 특성이 있습니다.

**Plotly Express 버전 코드**

```python
z = [[1, 20, 30],
    [20, 1, 60],
    [30, 60, 1]]
fig = px.imshow(z, color_continuous_scale='Viridis')
fig.show()
```

**Graph Object 버전 코드**

```python
fig = go.Figure(data=go.Heatmap(
                z=[[1, 20, 30],
                [20, 1, 60],
                [30, 60, 1]]))
fig.show()
```

## 8. 타임라인 차트 / 간트 차트

간트 차트란, 프로젝트 관리에서 각 작업의 일정과 진행 상황을 시각적으로 나타내는 도구입니다. 세로축에는 작업 또는 활동 목록이, 가로축에는 시간 축이 표시되며, 각 작업의 시작과 끝 시간을 막대 형태로 표현합니다.

Graph Object로는 간트 차트를 그리는 전용 메서드는 없습니다.

```python
df = pd.DataFrame([
dict(Task="Job A", Start='2009-01-01', Finish='2009-02-28', Resource="Alex"),
dict(Task="Job B", Start='2009-03-05', Finish='2009-04-15', Resource="Alex"),
dict(Task="Job C", Start='2009-02-20', Finish='2009-05-30', Resource="Max")
])

fig = px.timeline(df, x_start="Start", x_end="Finish", y="Resource", color="Resource")
fig.show()
```

## 9. FUNNEL 차트

프로세스의 각 단계에서 수치가 점차 줄어드는 과정을 시각적으로 나타내는 차트입니다. 보통 위에서 아래로 좁아지는 형태로 표현되며, 각 단계에서의 전환율, 이탈률 등을 한눈에 파악할 수 있습니다.

**Plotly Express 버전 코드**

```python
data = dict(
number=[39, 27.4, 20.6, 11, 2],
stage=["Website visit", "Downloads", "Potential customers", "Requested price", "Invoice sent"])
fig = px.funnel(data, x='number', y='stage')
fig.show()
```

**Graph Object 버전 코드**

```python
fig = go.Figure(go.Funnel(
y = ["Website visit", "Downloads", "Potential customers", "Requested price", "invoice sent"],
x = [39, 27.4, 20.6, 11, 2]))
fig.show()
```

## 10. 지도

지도는, 지리적 데이터를 시각화하여 각 지역별 정보를 한눈에 보여주는 차트 유형입니다. 지도 위에 데이터 포인트, 색상, 마커 등을 겹쳐 표현하여 지역별 특성이나 분포를 시각화합니다.

```python
df = px.data.carshare()
fig = px.scatter_map(df, lat="centroid_lat", lon="centroid_lon", color="peak_hour", size="car_hours",
color_continuous_scale=px.colors.cyclical.IceFire, size_max=15, zoom=10,
map_style="carto-positron")
fig.show()
```
