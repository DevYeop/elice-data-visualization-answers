## 1. 라이브러리 불러오기

먼저, 필요한 라이브러리들을 import를 통해 불러옵니다. 이론 시간에서 알아보았듯, Plotly에는 2개의 주요 모듈이 존재합니다.

각각 `Express`는 `px`라는 이름으로, `Graph Object`는 `go` 라는 이름으롭 사용하겠습니다.

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
```

## 2. Figure 구조 이해하기

이제 본격적으로 Plotly의 Figure 구조를 이해해보겠습니다. 

`Figure`는 전체 그래프를 나타내는 최상위 객체로, 그래프를 그리기 위한 캔버스라고 할 수 있습니다.

Figure는 크게 3가지 속성으로 구성되는데, `data`, `layout`, `frames` 입니다.

- `data` : 그래프의 타입(산점도, 막대그래프 등)을 정의
- `layout` : 축, 제목, 색상과 같은 그래프의 시각 요소를 설정
- `frames` : 애니메이션을 위한 각 프레임을 제어

```python
fig = px.line(x=["a","b","c"], y=[1,3,2], title="sample figure")
# print(fig)
fig
```

## 3. 그래프 그리기

아래 모양의 막대 그래프를 3가지 방법으로 그려보도록 하겠습니다. 

1) Dictionary 활용
2) Graph Object 활용
3) Plotly Express 활용

![image.png](./sample_graph.png)

### 1) Dictionary 활용

하나의 막대 (bar) Trace와 제목을 포함한 Figure를 딕셔너리 형태로 생성합니다. 

이 때, Figure을 구성하는 모든 속성과 데이터를 키-값 쌍으로 설정해야 하고, 만약 잘못된 속성명이나 속성 값을 사용하더라도 오류가 발생하지 않습니다.

Figure 구조를 직접 제어하고 수정할 수 있는 유연성이 있지만 동시에 모든 설정을 수동으로 해주어야 하기에 다소 비효율적인 방법이라 할 수 있습니다.

```python
# 🦁 plotly는 그래프를 그리기 위해 figure라는 객체를 사용
# - `data` : 그래프의 타입(산점도, 막대그래프 등)을 정의
# - `layout` : 축, 제목, 색상과 같은 그래프의 시각 요소를 설정

fig = dict({
    "data": [{"type": "bar",
              "x": [1, 2, 3],
              "y": [1, 3, 2]}],
    "layout": {"title": {"text": "A Figure Specified By Python Dictionary"}}
})


# 딕셔너리로 정의된 그래프를 출력하려면, 저수준 plotly.io.show 함수 사용 필요
import plotly.io as pio

pio.show(fig)
```

### 2) Graph Object 활용

`plotly.graph_objects.Figure` 를 사용하여 Figure를 표현합니다. 

이 방법에서는 Figure를 생성한 후 이 Figure에 다양한 속성을 추가하는 방식으로 그래프를 그리게 됩니다.

```python
# 이전 코드와 비교 🦁
fig = dict({
    "data": [{"type": "bar",
              "x": [1, 2, 3],
              "y": [1, 3, 2]}],
    "layout": {"title": {"text": "A Figure Specified By Python Dictionary"}}
})
#################### ############################################################

fig = go.Figure(
    data=[go.Bar(x=[1, 2, 3], y=[1, 3, 2])],
    layout=go.Layout(
        title=go.layout.Title(text="A Figure Specified By A Graph Object")
    )
)

# plotly.io.show()를 사용하지 않고, fig.show()로 바로 호출함으로써,
# 보다 간편히 호출이 가능하다.
fig.show()
```

**Graph Object로 그래프를 그리는 또 다른 방법**

이 때, 앞 Dictionary를 활용한 방법과 결합하면, 아래와 같이 그래프를 그릴 수 있습니다. 

```python

import plotly.graph_objects as go

# 이전 코드와 비교 🦁
fig = dict({
    "data": [{"type": "bar",
              "x": [1, 2, 3],
              "y": [1, 3, 2]}],
    "layout": {"title": {"text": "A Figure Specified By Python Dictionary"}}
})

import plotly.io as pio

# pio.show(fig)
print(type(fig))
# <class 'dict'>
################################################################################

fig = dict({
    "data": [{"type": "bar",
    "x": [1, 2, 3],
    "y": [1, 3, 2]}],
    "layout": {"title": {"text": "A Figure Specified By A Graph Object With A Dictionary"}}
})

fig = go.Figure(fig)
fig.show()
print(type(fig))
# <class 'plotly.graph_objects.Figure'>
# Figure 객체는 plotly에서 그래프를 그리기 위해 사용하는 객체로, data와 layout 속성을 포함하고 있다.
# 해당 객체를 사용하면 그래프를 그리기 위한 다양한 설정을 할 수 있으며, plotly의 다양한 기능을 활용할 수 있다.⭐
```

Dictionary만 사용했을 때와는 달리, 잘못된 속성명이나 속성 값을 사용할 경우 오류 메세지가 표시됩니다. 

```python
fig = dict({
    "data": [{"type": "bar",
    "x": [1, 2, 3],
    "y": [1, 3, 2]}],
    "layout": {"titl": {"text": "A Figure Specified By A Graph Object With A Dictionary"}}
})

try:
    go.Figure(fig)
except ValueError:
    print("Plotly에서 titl은 유효한 속성이 아닙니다. title을 사용하세요.")
```

**Graph Object를 딕셔너리로 변환하기**

Graph Object를 `to_dict()` 메서드를 사용하여 딕셔너리 형식으로 변환하여 출력할 수도 있습니다. 

이 방법을 통해 Figure의 구조를 확인할 수 있습니다.

```python
fig = dict({
    "data": [{"type": "bar",
    "x": [1, 2, 3],
    "y": [1, 3, 2]}],
    "layout": {"title": {"text": "A Figure Specified By A Graph Object With A Dictionary"}}
})
fig = go.Figure(fig) 
#### 위의 코드는 오탈자가 없는 상태로, 정상적으로 그래프 객체로 변환이 가능하다.
#### 위 셀의 오류를 정정하기 위해 추가삽입된 코드 🦁🦁

fig.layout.template = None # 출력  결과 축소하기. 그냥 data외 내용을 지워서 콘솔출력을 줄이려고 적힌 코드 

# Figure 객체를 딕셔너리로 변환하면, data와 layout 속성을 포함한 딕셔너리 형태로 변환된다.
print("Dictionary Representation of A Graph Object:\n\n" + str(fig.to_dict()))
```

## 3. Ploty Express 활용

단일 함수 호출만으로도 그래프 객체 Figure를 생성할 수 있습니다. 

이 때는, Graph Object와 다르게 한 번의 함수 호출로도 그래프를 생성할 수 있게 됩니다. 

```python
import plotly.express as px

# 데이터 생성
data = {
    "x": [1, 2, 3],
    "y": [1, 3, 2]
}

# 그래프 생성
fig = px.bar(data, x="x", y="y", title="A Figure Specified By Plotly Express")

# 그래프 표시3
fig.show()
```

앞 Graph Object 방법과 비교했을 때 비교적 간단하고 짧은 코드 만으로도 그래프를 그릴 수 있는 것을 볼 수 있습니다. 

특히, Plotly Express에는 30가지 이상의 그래프를 생성할 수 있는 함수를 제공하여 다양한 유형의 그래프를 쉽게 그릴 수 있습니다.

### 그래프 그리기 방법 선택 가이드

그래프를 그릴 때는, 아래의 예시 코드와 같은 방법을 사용하는 것을 권장합니다. 

1) Plotly Express를 사용하여 Figure을 한 번에 생성한 다음, 

2) plotly.graph_objects.Figure 객체를 수정하여 필요 시, 복잡한 수동 설정 진행

```python
# 1) Plotly Express를 사용하여 기본 산포도 생성
df = px.data.gapminder().query("year == 2007")
fig = px.scatter(df, x="gdpPercap", y="lifeExp", color="continent", size="pop", hover_name="country", log_x=True)

# 2) plotly.graph_objects.Figure 객체를 사용하여 추가 설정 진행
# - 레이아웃 변경
fig.update_layout(
title="2007년 각 대륙의 GDP 대비 기대수명",
xaxis_title="1인당 GDP (log scale)",
yaxis_title="기대수명",
legend_title="대륙",
title_font_size=20
)

# - 각 데이터 포인트(마커)에 테두리 추가
fig.update_traces(marker=dict(line=dict(width=2, color='DarkSlateGrey')))

# - x, y축 눈금 및 레이아웃 배경색 설정
fig.update_xaxes(showgrid=True, gridwidth=1, gridcolor='LightGrey')
fig.update_yaxes(showgrid=True, gridwidth=1, gridcolor='LightGrey')
fig.update_layout(plot_bgcolor="LightBlue")

# 그래프 출력하기
fig.show()
```
