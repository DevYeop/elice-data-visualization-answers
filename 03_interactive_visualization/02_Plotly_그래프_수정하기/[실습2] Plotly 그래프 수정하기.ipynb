## 1. 그래프 추가하기

`add_trace()` 메서드를 사용하여 Figure에 그래프를 추가할 수 있습니다. 

이 때, `add_trace()` 메서드는 추가할 Trace를 인수로 받아 Figure 에 추가합니다. 
우리는 이 메서드를 통해 처음 빈 상태로 생성된 Figure에 다양한 Trace를 추가할 수 있게 됩니다.

```python
fig = go.Figure()
fig.add_trace(go.Bar(x=[1, 2, 3], y=[1, 3, 2]))
fig.show()
```

Graph Object로 생성된 Figure 외에도 Express로 생성된 Figure에도 동일하게 적용할 수 있습니다. 

```python
df = px.data.iris()
# 🦁 print(df)

# 🦁 color="species"는 색상 자체를 지정하는 것이 아니라,
# 🦁 species 컬럼의 값(setosa, versicolor, virginica)에 따라
# 🦁 데이터 포인트의 색상을 자동으로 구분하는 역할을 한다.
# 🦁 즉, color="red"처럼 특정 색상을 지정하는 것과는 다르다.
fig = px.scatter(df, x="sepal_width", y="sepal_length", color="species",
                 title="Using The add_trace() method With A Plotly Express Figure")

fig.add_trace(
    go.Scatter(
        x=[2, 4],
        y=[4, 8],
        mode="lines",
        line=go.scatter.Line(color="red"),
        showlegend=False)
)
fig.show()
```

혹은 각 Trace 별로 제공하는 `add_{trace}()` 메서드를 사용해서 Figure에 그래프를 추가할수도 있습니다. 

```python
# 데이터 생성
data = {
    "x": [1, 2, 3],
    "y": [1, 3, 2]
}

# 그래프 생성
fig = px.bar(data, x="x", y="y", title="A Figure Specified By Plotly Express")

# 각 막대 위를 점으로 이어주는 선 추가
fig.add_scatter(x=data["x"], y=data["y"], mode="lines+markers", name="Line Over Bars")
# 라인을 빼면 실제로 산점도가 추가된 것처럼 보인다. 메서드명이 이상한게 아니다 🦁
# fig.add_scatter(x=data["x"], y=data["y"], mode="markers", name="Line Over Bars")

# 그래프 표시
fig.show()
```

## 2. 그래프 레이아웃 수정하기

Figure의 그래프 외에도 그래프의 레이아웃 (제목 크기, 색상 등)을 변경할 수도 있습니다.

이 때는, `update_layout()` 메서드를 활용합니다. 
이 때, 아래와 같이 여러 속성을 쉽게 설정할 수 있도록 매직 언더스코어(`_`) 표기법을 사용할 수 있습니다. 

```python
fig = go.Figure(data=go.Bar(x=[1, 2, 3], y=[1, 3, 2]))
fig.update_layout(title_text="Using update_layout() With Graph Object Figures", title_font_size=30)
# _를 활용하여 속성을 보다 짧게 표현함 
# 원래는 아래와 같이 써야한다 💀
# fig.update_layout(
#     title=dict(
#         text="그래프 제목",
#         font=dict(size=30)
#     )
# )
##############

"""
fig.update_layout(
    title=dict(
        text="Using update_layout() With Graph Object Figures",
        font=dict(size=30)
 )
)
"""

fig.show()
```

아래는 동일한 속성을 적용하는 다양한 방법을 제시합니다.

```python
# _ 가 .대신에 붙었다고 생각하면 된다. 즉, title_text는 title.text를 의미하고, title_font_size는 title.font.size를 의미한다.

fig.update_layout(title_text="update_layout() Syntax Example",
                  title_font_size=30)

fig.update_layout(title_text="update_layout() Syntax Example",
                  title_font=dict(size=30))


fig.update_layout(title=dict(text="update_layout() Syntax Example"),
                             font=dict(size=30))

fig.update_layout({"title": {"text": "update_layout() Syntax Example",
                             "font": {"size": 30}}})

fig.update_layout(title=go.layout.Title(text="update_layout() Syntax Example",
                                        font=go.layout.title.Font(size=30)))
```

## 3. 한 도화지에 여러 그래프 그리기

Matplotlib, Seaborn과 마찬가지로 Plotly에서도 1개의 Figure에 여러 개의 그래프를 그릴 수 있습니다. 

이 때는, `plotly.subplots`의 `make_subplots` 를 활용합니다. 

```python
from plotly.subplots import make_subplots
fig = make_subplots(rows=1, cols=2)
fig.add_trace(go.Scatter(y=[4, 2, 1], mode="lines"), row=1, col=2) # 🦁뒤에는 포지션, 바꿔주면 표시위치가 바뀐다
fig.add_trace(go.Bar(y=[2, 1, 3]), row=1, col=1)
fig.show()
```

혹은, Express에서는 `facet_col` 혹은 `facet_row` 속성을 활용하여 여러 개의 그래프를 그릴 수 있습니다.

```python
df = px.data.iris()

# 🦁 facet_col을 사용하여 species별로 서브플롯을 생성하고, 각 서브플롯에 산점도를 그린다.
fig = px.scatter(df, x="sepal_width", y="sepal_length", color="species", facet_col="species",
                 title="Adding Traces To Subplots Witin A Plotly Express Figure")

# 🦁 각 서브플롯에 동일한 참조선을 추가하기 위해 go.Scatter 객체를 생성한다.
reference_line = go.Scatter(x=[2, 4],
                            y=[4, 8],
                            mode="lines",
                            line=go.scatter.Line(color="gray"),
                            showlegend=False)

# 🦁 그리고 포지션을 지정하여 각 서브플롯에 참조선을 추가한다.
fig.add_trace(reference_line, row=1, col=1)
fig.add_trace(reference_line, row=1, col=2)
fig.add_trace(reference_line, row=1, col=3)

fig.show()
```

## 4. 그 외 그래프 수정 메서드

그 외에도 `update_traces()`, `for_each_trace()`, `update_xaxes()`, `update_yaxes()` 와 같은 메서드를 사용하여 다양한 속성을 변경할 수 있습니다. 

### 1. `update_traces()`

여러 개의 Traces들의 속성을 동시에 변경할 수 있습니다.

```python
fig = make_subplots(rows=1, cols=2)

fig.add_scatter(y=[4, 2, 3.5], mode="markers",
                marker=dict(size=20, color="LightSeaGreen"),
                name="a", row=1, col=1)

fig.add_bar(y=[2, 1, 3],
            marker=dict(color="MediumPurple"),
            name="b", row=1, col=1)

fig.add_scatter(y=[2, 3.5, 4], mode="markers",
                marker=dict(size=20, color="MediumPurple"),
                name="c", row=1, col=2)

fig.add_bar(y=[1, 3, 2],
            marker=dict(color="LightSeaGreen"),
            name="d", row=1, col=2)

fig.show()
```

```python
fig = make_subplots(rows=1, cols=2)

fig.add_scatter(y=[4, 2, 3.5], mode="markers",
                marker=dict(size=20, color="LightSeaGreen"),
                name="a", row=1, col=1)

fig.add_bar(y=[2, 1, 3],
            marker=dict(color="MediumPurple"),
            name="b", row=1, col=1)

fig.add_scatter(y=[2, 3.5, 4], mode="markers",
                marker=dict(size=20, color="MediumPurple"),
                name="c", row=1, col=2)

fig.add_bar(y=[1, 3, 2],
            marker=dict(color="LightSeaGreen"),
            name="d", row=1, col=2)

fig.update_traces(marker=dict(color="RoyalBlue"),
                  col=2)

fig.show()
```

### 2. `for_each_trace()`

만약, 여러 개의 Trace에 적용하려는 변경사항이 특정 Trace의 속성의 현재 값에 따라 달라질 경우, 사용하는 메서드입니다.

```python
df = px.data.iris()

fig = px.scatter(df, x="sepal_width", y="sepal_length", color="species",
                 title="Conditionally Updating Traces In A Plotly Express Figure With for_each_trace()")

fig.for_each_trace(
    lambda trace: trace.update(marker_symbol="square") if trace.name == "setosa" else (),
)

fig.show()
```

### 3. `update_xaxes()` / `update_yaxes()`

x 축과 y 축의 속성 값을 변경할 때 사용하는 메서드입니다.

```python
df = px.data.iris()

fig = px.scatter(df, x="sepal_width", y="sepal_length", color="species",
                 facet_col="species", title="Using update_xaxes() With A Plotly Express Figure")

fig.update_xaxes(showgrid=False)

fig.show()
```
