## 1. 테마 설정 기능

Plotly에서는 Seaborn 처럼 다양한 테마를 사전에 지정해두고, 이를 사용자가 쉽게 설정할 수 있도록 합니다.

아래에서는 총 7개 (none - 테마 비활성화 포함)의 테마를 지정하여 각 테마를 비교해보도록 하겠습니다.

```python
import plotly.express as px
df = px.data.gapminder()
df_2007 = df.query("year==2007")
for template in ["plotly", "plotly_white", "plotly_dark", "ggplot2", "seaborn", "simple_white", "none"]:
    fig = px.scatter(df_2007,
                     x="gdpPercap", y="lifeExp", size="pop", color="continent",
                     log_x=True, size_max=60,
                     template=template, title="Gapminder 2007: '%s' theme" % template)
    fig.show()
```

이러한 테마를 전체적으로 적용하기 위해 default 템플릿을 지정할수도 있습니다.

해당 템플릿 지정 이후에는 모든 그래프가 설정된 테마로 그려지게 됩니다.

```python
import plotly.io as pio
import plotly.express as px

pio.templates.default = "plotly_white"

df = px.data.gapminder()
df_2007 = df.query("year==2007")

fig = px.scatter(df_2007,
                 x="gdpPercap", y="lifeExp", size="pop", color="continent",
                 log_x=True, size_max=60,
                 title="Gapminder 2007: current default theme")
fig.show()
```

원한다면 아래와 같이 직접 테마를 생성하고, 이를 전체적으로 적용할수도 있습니다.

```python
import plotly.graph_objects as go

large_rockwell_template = dict(
    layout=go.Layout(title_font=dict(family="Rockwell", size=24))
)

fig = go.Figure()
fig.update_layout(title="Figure Title",
                  template=large_rockwell_template)
fig.show()
```

```python
import plotly.graph_objects as go

draft_template = go.layout.Template()
draft_template.layout.annotations = [
    dict(
        name="draft watermark",
        text="Confidential",
        textangle=-30,
        opacity=0.1,
        font=dict(color="black", size=100),
        xref="paper",
        yref="paper",
        x=0.5,
        y=0.5,
        showarrow=False,
    )
]

fig=go.Figure()
fig.update_layout(template=draft_template)
fig.show()
```

```python
df = px.data.gapminder()
df_2007 = df.query("year==2007")

fig = px.scatter(df_2007,
                 x="gdpPercap", y="lifeExp", size="pop", color="continent",
                 log_x=True, size_max=60,
                 title="Gapminder 2007: current default theme")
fig.update_layout(template=draft_template)
fig.show()
```

## 2. 한글 폰트 설정 기능

앞의 Matplotlib과 동일하게 Plotly에서도 한글 폰트를 그래프에 출력하기 위해서는 별도의 한글 폰트 설정이 필요합니다.

이 방법은 위 디폴트 템플릿을 지정하는 방법과 유사합니다.

```python
# 그래프 한글 사용을 위한 전역 레이아웃 설정 (폰트를 'Malgun Gothic'으로 설정)

pio.templates["custom"] = pio.templates["plotly"] # 기본 템플릿 복사
pio.templates["custom"]["layout"]["font"] = dict(family="Malgun Gothic", size=16)

pio.templates.default = "custom"
```

## 3. 대화형 기능 설정

다양한 대화형 기능을 `show()` 메서드의 `config` 매개변수를 통해 적용할 수 있습니다. 

먼저, 스크롤 줌(확대) 기능을 추가해보겠습니다.

```python
import plotly.graph_objects as go
fig = go.Figure()
config = {'scrollZoom': True}

fig.add_trace(
go.Scatter(
x=[1, 2, 3],
y=[1, 3, 1]))

fig.show(config=config)
```

기본적으로 Plotly에 생성된 Figure는 반응형 그래프입니다. 

만약, 이 그래프 설정을 비활성화하고 싶다면 아래와 같이 적용할 수 있습니다.

창의 크기를 변경해보며 반응형 기능이 적절하게 비활성화되었는지 확인해보세요.

```python
config = {'responsive': False}

fig.show(config=config)
```

그 외 다양한 설정들이 존재합니다.

```python
config = {'staticPlot': False,
          'displayModeBar': True,
          'displaylogo': False,
          'toImageButtonOptions': {
              'format': 'svg', # one of png, svg, jpeg, webp
              'filename': 'custom_image',
              'height': 500, # None : 
              'width': 700, # None : 
              'scale': 1, # Multiply title/legend/axis/canvas sizes by this factor
          'modeBarButtonsToRemove': ['zoom', 'pan']}
         }
fig.show(config=config)
```

```python
fig.show(config={'modeBarButtonsToAdd': ['drawline',
    'drawopenpath',
    'drawclosedpath',
    'drawcircle',
    'drawrect',
    'eraseshape'
]})
```

## 4. 마우스 호버

마우스 호버란, 마우스를 특정 요소 위에 올려놓는 동작을 의미합니다. 사용자가 마우스를 그래프, 버튼, 이미지 등의 요소 위에 올려놓을 때 추가적인 정보나 효과를 제공하는 UI 상호작용 방식을 간단하게 적용할 수 있습니다.

```python
df = px.data.gapminder()
fig = px.scatter(df.query("year==2007"), x="gdpPercap", y="lifeExp", size="pop", color="continent",
hover_name="country", log_x=True, size_max=60)
fig.show()
```

## 5. 애니메이션

Plotly에서 애니메이션은 시간의 흐름이나 데이터의 변화를 시각적으로 표현하여 데이터를 더욱 직관적으로 이해할 수 있게 해줍니다. 

이러한 애니메이션을 통해 데이터가 여러 프레임에 걸쳐 변화하는 모습을 표시할 수 있어, 특히 시계열 데이터나 상태 변화 분석에 유용합니다.

```python
df = px.data.gapminder()
fig = px.scatter(df, x="gdpPercap", y="lifeExp", animation_frame="year", animation_group="country",
size="pop", color="continent", hover_name="country", facet_col="continent",
log_x=True, size_max=45, range_x=[100,100000], range_y=[25,90])
fig.show()
```
